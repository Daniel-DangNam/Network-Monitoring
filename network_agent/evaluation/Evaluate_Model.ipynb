```python
import os
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torchvision.models as models
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix
import warnings
warnings.filterwarnings('ignore')

# Cài đặt giao diện biểu đồ
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 10)

# --- ĐỊNH NGHĨA 11 NHÃN CHUẨN CỦA MẠNG RESNET-18 ---
CLASS_NAMES = [
    'Streaming', 'VPN-VoIP', 'VPN-chat', 'VPN-email', 'VPN-file_transfer',
    'VPN-p2p', 'VPN-streaming', 'VoIP', 'chat', 'email', 'file_transfer'
]

# --- BỘ TỪ ĐIỂN RULE-BASED THỰC CHIẾN (CHUẨN HÓA THEO PARALLEL FUSION) ---
SNI_RULES_ADVANCED = {
    # 1. EMAIL
    'mail.google.com': 'email',
    'outlook.live.com': 'email',
    'outlook.office.com': 'email',
    'outlook.office365.com': 'email',
    'smtp': 'email',
    'imap': 'email',
    'pop3': 'email',

    # 2. STREAMING
    'googlevideo.com': 'Streaming',
    'youtube.com': 'Streaming',
    'ytimg.com': 'Streaming',
    'video.xx.fbcdn.net': 'Streaming',
    'nflxvideo.net': 'Streaming',
    'netflix.com': 'Streaming',
    'vimeocdn.com': 'Streaming',
    'ttvnw.net': 'Streaming',
    
    # 3. FILE TRANSFER
    'drive.google.com': 'file_transfer',
    'docs.google.com': 'file_transfer',
    'dropbox.com': 'file_transfer',
    'onedrive.live.com': 'file_transfer',
    'fbsbx.com': 'file_transfer',
    'sharepoint.com': 'file_transfer',
    'mediafire.com': 'file_transfer',
    'mega.nz': 'file_transfer',
}
```

```python
def load_ai_model(model_path):
    print("[AI] Đang nạp cấu trúc ResNet-18 (Chế độ Đánh giá Lab)...")
    model = models.resnet18(weights=None)
    model.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
    num_ftrs = model.fc.in_features
    model.fc = nn.Sequential(
        nn.Dropout(0.5),
        nn.Linear(num_ftrs, 11)
    )
    checkpoint = torch.load(model_path, map_location=torch.device('cpu'))
    if 'model_state_dict' in checkpoint:
        model.load_state_dict(checkpoint['model_state_dict'])
    else:
        model.load_state_dict(checkpoint)
    model.eval()
    print("[AI] Lõi AI đã sẵn sàng!")
    return model

def features_to_tensor(features_list):
    features_81 = list(features_list)
    if len(features_81) < 81:
        features_81.extend([0.0] * (81 - len(features_81)))
    elif len(features_81) > 81:
        features_81 = features_81[:81]
    
    arr = np.array(features_81, dtype=np.float32)
    arr = np.nan_to_num(arr, nan=0.0, posinf=1.0, neginf=0.0)
    
    arr_min, arr_max = np.min(arr), np.max(arr)
    if arr_max - arr_min != 0:
        arr_normalized = (arr - arr_min) / (arr_max - arr_min)
    else:
        arr_normalized = arr - arr_min
        
    image_9x9 = arr_normalized.reshape(9, 9)
    image_rgb = np.stack([image_9x9, image_9x9, image_9x9], axis=0)
    return torch.tensor(np.expand_dims(image_rgb, axis=0), dtype=torch.float32)

def parallel_fusion_predict_eval(model, tensor_input, sni_domain, protocol, dst_port=0):
    domain_lower = str(sni_domain).lower().strip() if sni_domain else ""
    is_anonymous = domain_lower in ['', 'none', 'unknown', 'nan']
    
    # ==========================================
    # BÁC SĨ 1: RULE-BASED
    # ==========================================
    rule_pred = "UNKNOWN"
    
    # 1. BẮT GIAO THỨC TRƯỚC (CỨU CÁNH CHO WEBRTC / GOOGLE MEET ẨN DANH)
    if protocol == 17 and (dst_port == 3478 or dst_port > 10000 or (19302 <= dst_port <= 19309)):
        rule_pred = 'VoIP'
        
    # 2. KIỂM TRA TÊN MIỀN NẾU KHÔNG BỊ ẨN DANH
    elif not is_anonymous:
        if 'googleusercontent.com' in domain_lower:
            subdomain = domain_lower.split('.googleusercontent.com')[0]
            if subdomain.startswith('ci') or 'mail' in subdomain or subdomain.startswith('lh'):
                rule_pred = 'email'
            else:
                rule_pred = 'google_ecosystem'
        elif any(app in domain_lower for app in ['meet.google', 'zoom', 'teams', 'webex']):
            rule_pred = 'VoIP'
        elif any(app in domain_lower for app in ['zalo', 'zadn.vn', 'telegram', 'messenger', 'discord', 'skype']):
            rule_pred = 'chat'
            if protocol == 17 or dst_port == 3478: 
                rule_pred = 'VoIP'
        else:
            for key, label in SNI_RULES_ADVANCED.items():
                if key in domain_lower:
                    rule_pred = label
                    break

    # ==========================================
    # BÁC SĨ 2: AI RESNET-18
    # ==========================================
    with torch.no_grad():
        output = model(tensor_input)
        probabilities = torch.softmax(output, dim=1)
        confidence, idx = torch.max(probabilities, 1)
        
        ai_pred = CLASS_NAMES[idx.item()]
        ai_conf = confidence.item()

    # ==========================================
    # TRỌNG TÀI DÀNH RIÊNG CHO ĐÁNH GIÁ CSV (ÉP BUỘC DỰ ĐOÁN)
    # ==========================================
    if rule_pred == ai_pred:
        return rule_pred
        
    if rule_pred == 'google_ecosystem':
        if ai_pred in ['file_transfer', 'email']:
            return ai_pred
        return 'file_transfer'
        
    # Áp dụng AI Confidence > 0.60 giống hệt logic thực tế
    if rule_pred == 'chat' and ai_pred == 'VoIP' and ai_conf > 0.60: 
        return 'VoIP'
        
    if rule_pred != 'UNKNOWN':
        return rule_pred
        
    # NẾU CSV KHÔNG CÓ TÊN MIỀN VÀ KHÔNG PHẢI WEBRTC, ÉP AI PHẢI QUYẾT ĐỊNH
    final_pred = ai_pred.replace("VPN-", "")
    
    if final_pred == 'streaming':
        return 'Streaming'
    if final_pred == 'p2p':
        return 'IGNORED'
        
    return final_pred
```

```python
# THAY ĐỔI ĐƯỜNG DẪN BÊN DƯỚI CHO ĐÚNG VỚI MÁY CỦA BẠN NẾU CẦN
model_path = "model_final_Application_Label.pth" 
model = load_ai_model(model_path)
```

```python
import os
import pandas as pd

# Cấu hình map 5 file với đúng nhãn thực tế (Đã giữ nguyên đường dẫn tuyệt đối của bạn)
files_mapping = {
    "/home/namdang/Desktop/Network_Monitoring/network_agent/csv/01_Sreaming.csv": "Streaming",
    "/home/namdang/Desktop/Network_Monitoring/network_agent/csv/02_File-transfer.csv": "file_transfer",
    "/home/namdang/Desktop/Network_Monitoring/network_agent/csv/03_Email.csv": "email",
    "/home/namdang/Desktop/Network_Monitoring/network_agent/csv/04_VoIP.csv": "VoIP",
    "/home/namdang/Desktop/Network_Monitoring/network_agent/csv/05_Chat.csv": "chat"
}

dataframes = []

# Duyệt thẳng qua đường dẫn tuyệt đối
for file_path, true_label in files_mapping.items():
    if os.path.exists(file_path):
        df = pd.read_csv(file_path)
        
        # Chỉ lọc bỏ các dòng bị lỗi vật lý (Error), GIỮ LẠI toàn bộ các dòng để ép mô hình đánh giá
        if 'AI_Prediction' in df.columns:
            df = df[df['AI_Prediction'] != 'Error']
        
        y_pred_new = []
        # Lấy tên file ngắn gọn để in ra màn hình cho đẹp
        short_name = os.path.basename(file_path)
        print(f"Đang đánh giá file: {short_name} (True Label: {true_label})...")
        
        for index, row in df.iterrows():
            domain = row.get('Destination / Websites', 'Unknown')
            protocol = row.get('Protocol', 6) if 'Protocol' in df.columns else row.get('protocol', 6)
            dst_port = row.get('Dst Port', 0) if 'Dst Port' in df.columns else row.get('dst_port', 0)
            
            cols_to_drop = ['ID', 'Timestamp', 'Destination / Websites', 'AI_Prediction', 'True_Label', 'Flow ID', 'Src IP', 'Dst IP']
            numeric_cols = [c for c in df.columns if c not in cols_to_drop and df[c].dtype != 'object']
            
            features = row[numeric_cols].tolist()
            tensor_data = features_to_tensor(features)
            
            # ---> QUAN TRỌNG: GỌI ĐÚNG HÀM ÉP BUỘC ĐÁNH GIÁ (EVALUATION MODE) <---
            new_pred = parallel_fusion_predict_eval(model, tensor_data, domain, protocol, dst_port)
            y_pred_new.append(new_pred)
        
        df['AI_Prediction_NEW'] = y_pred_new
        df['True_Label'] = true_label
        dataframes.append(df)
        print(f"✅ Hoàn tất file: {short_name} ({len(df)} luồng)")
    else:
        print(f"⚠️ Không tìm thấy file: {file_path}")

if dataframes:
    df_eval = pd.concat(dataframes, ignore_index=True)
    print(f"\n=> TỔNG SỐ LUỒNG DỮ LIỆU ĐÁNH GIÁ THỰC TẾ: {len(df_eval)}")
    
    y_true = df_eval['True_Label']
    y_pred = df_eval['AI_Prediction_NEW']
else:
    print("Không có dữ liệu để đánh giá.")
```

```python
if 'df_eval' in locals() and not df_eval.empty:
    # Cố định 5 nhãn ứng dụng cốt lõi để báo cáo không bị rác bởi các nhãn nhiễu
    target_labels = ['Streaming', 'VoIP', 'chat', 'email', 'file_transfer']
    
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, average='macro', zero_division=0)
    recall = recall_score(y_true, y_pred, average='macro', zero_division=0)
    f1 = f1_score(y_true, y_pred, average='macro', zero_division=0)

    print("="*65)
    print(" KẾT QUẢ ĐÁNH GIÁ TỔNG THỂ KIẾN TRÚC DUNG HỢP (PARALLEL FUSION)")
    print("="*65)
    print(f" Accuracy  (Độ chính xác) : {accuracy:.4f} ({accuracy*100:.2f}%)")
    print(f" Precision (Độ chuẩn xác) : {precision:.4f} ({precision*100:.2f}%)")
    print(f" Recall    (Độ bao phủ)   : {recall:.4f} ({recall*100:.2f}%)")
    print(f" F1-Score  (Điểm cân bằng): {f1:.4f} ({f1*100:.2f}%)")
    print("="*65)

    print("\n BÁO CÁO CHI TIẾT THEO TỪNG HÀNH VI (5 NHÃN CHUẨN):")
    # Ép classification_report chỉ chấm điểm trên 5 nhãn cốt lõi
    print(classification_report(y_true, y_pred, labels=target_labels, zero_division=0))
    
    # Bộ đếm cảnh báo nhãn rác (Phục vụ debug)
    unique_preds = set(y_pred)
    stray_labels = unique_preds - set(target_labels)
    if stray_labels:
        print(f"\n⚠️ [CẢNH BÁO DEBUG]: Trọng tài Fusion đã đẩy một số luồng vào các nhãn ngoài dự kiến: {stray_labels}")
        print("   -> Điều này xảy ra do có dữ liệu nhiễu không xác định, hoặc luồng mã hóa bị ép thành IGNORED.")
else:
    print("Không có dữ liệu hợp lệ để tiến hành đánh giá.")
```

```python
if 'df_eval' in locals() and not df_eval.empty:
    # 1. Khóa danh sách 5 nhãn cốt lõi lên đầu để ma trận luôn đẹp và nhất quán
    target_labels = ['Streaming', 'VoIP', 'chat', 'email', 'file_transfer']
    
    # 2. Gom thêm các nhãn lạ/nhãn rác (nếu có) xuống cuối ma trận để dễ debug
    actual_labels = set(y_true).union(set(y_pred))
    stray_labels = sorted(list(actual_labels - set(target_labels)))
    final_labels = target_labels + stray_labels

    cm = confusion_matrix(y_true, y_pred, labels=final_labels)

    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=final_labels, yticklabels=final_labels)

    # 3. Đổi tên tiêu đề phản ánh đúng Kiến trúc Dung hợp (Parallel Fusion)
    plt.title('Ma Trận Nhầm Lẫn (Confusion Matrix) - Parallel Fusion Engine', fontsize=16, fontweight='bold', pad=20)
    plt.xlabel('Nhãn Dự Đoán (Predicted Label)', fontsize=12, fontweight='bold')
    plt.ylabel('Nhãn Thực Tế (True Label)', fontsize=12, fontweight='bold')
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()

    # 4. Lưu ảnh với tên gọi chuẩn xác
    plt.savefig("Confusion_Matrix_Parallel_Fusion.png", dpi=300)
    plt.show()
else:
    print("Không có dữ liệu để vẽ biểu đồ.")
```
